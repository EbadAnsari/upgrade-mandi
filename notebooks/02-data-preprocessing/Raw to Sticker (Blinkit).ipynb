```python
import pandas as pd
from datetime import date, datetime, timedelta
import numpy as np
```

```python
item_id = pd.read_excel(
    "./../../upgrade_mandi/raw-sheets-dump/blinkit/UPGRADE UPC.xlsx",
    sheet_name="Sheet1",
)[["Item ID", "UPC"]]

raw_file = pd.read_excel(
    "./../../upgrade_mandi/raw-sheets-dump/blinkit/jpgtoexcel_table.xlsx",
    sheet_name="Sheet1",
)
```

```python
item_id
```

```python
raw_file["Item ID"]  =raw_file["Item ID"].apply(lambda x: 0 if str(x).lower().strip() == "nan" else int(x))
```

```python
newDf = raw_file.merge(item_id, how="inner", left_on="Item ID", right_on="Item ID")
```

```python
newDf
```

```python
newDf["Date"] = date.today().strftime("%d-%m-%Y")
newDf["Best Before Date"] = (datetime.now() + timedelta(days=6)).strftime("%d-%m-%Y")
```

```python
finalDf = newDf.loc[newDf.index.repeat(newDf["Final Indent (Qty) Dis"])][
    ["Item Name", "UOM", "Date", "Best Before Date", "UPC"]
].reset_index(drop=True)
```

```python
finalDf.to_excel("blinkit sticker.xlsx", index=False)
```
