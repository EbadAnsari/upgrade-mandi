```python
import pandas as pd

import numpy as np

from os import makedirs

from utils.utils import nameExtracter
```

```python
customer_details = pd.read_csv("./../data/processed/B2B/customer_details_rows.csv")
```

```python
customer_details.columns = customer_details.columns.str.title()
```

```python
transaction_list = pd.read_csv("./../data/processed/B2B/Transaction List.csv")
```

```python
transaction_list.columns = transaction_list.columns.str.title()
```

```python
orders_table = transaction_list.merge(
    customer_details,
    left_on="Name",
    right_on="Name",
    how="outer",
)[["Date", "Name", "Item_Name", "Quantity", "Rate", "Id"]]
```

```python
orders_table["Item_Name"] = orders_table["Item_Name"].str.strip()
```

```python
list(orders_table.groupby("Item_Name").groups.keys())
```

```python
orders_table.rename(columns={"Id": "Customer_Id"}, inplace=True)
```

```python
orders_table["key"] = orders_table.apply(
    lambda row: row["Date"] + row["Item_Name"] + row["Customer_Id"], axis=1
)
```

```python
orders_table.drop_duplicates(subset=["key"], inplace=True)
```

```python
orders_table.to_csv(
    "./../data/processed/B2B/orders.csv",
    index=False,
    header=list(orders_table.columns.str.lower()),
)
```
