```python
prev = 0


def pretty(label: str, completed: float, total: int, length: int = 30) -> None:
    global prev
    print(
        " " * (prev * 2)
        + f"\r{label} ["
        + "=" * int(completed / total * length)
        + "-" * int((total - completed) / total * length)
        + f"] {int(completed / total * 100)}%",
        end="",
    )
    prev = length


def prettyPrintReplace(label: str) -> None:
    global prev
    label = str(label)
    print((" " * (prev * 2)) + f"\r{label}", end="")
    prev = len(label)
```

```python
# Connect and login
mail = imaplib.IMAP4_SSL(IMAP_HOST)
mail.login(EMAIL_USER, EMAIL_PASS)
mail.select("inbox")

statusRaw, messagesRaw = mail.search(None, f'(SUBJECT "{subject_to_search}")')
```

```python
grnPrDF = pd.DataFrame(
    {
        "Vendor Code": [],
        "Vendor Name": [],
        "PO": [],
        "ASN": [],
        "Invoice Number": [],
        "Invoice Amount": [],
        "Debit Note Amount": [],
    }
)

status = statusRaw
messages = messagesRaw
total_mails = len(messages[0].split())
count = 0
if status == "OK":
    for num in messages[0].split():
        typ, data = mail.fetch(num, "(RFC822)")
        msg = email.message_from_bytes(data[0][1])
        sub = msg["Subject"]

        for part in msg.walk():
            content_type = part.get_content_type()

            if content_type == "text/html":
                raw_html_text = part.get_payload(decode=True).decode()
                processed_dom_tree = BeautifulSoup(raw_html_text, "html.parser")
                table = processed_dom_tree.find("table")
                df = pd.read_html(str(table))[0]
                df = df.T
                df.drop(0)
                df.columns = grnPrDF.columns
                grnPrDF = pd.concat([grnPrDF, df.iloc[1:]])
                count += 1
                print(count, total_mails)
                break
```
