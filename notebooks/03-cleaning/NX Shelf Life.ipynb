```python
import pandas as pd
```

```python
sl = pd.read_excel(
    "./../../upgrade_mandi/shelf-life/Summer 2026 PUNE City Shelf Life.xlsx", "Sheet1"
)
sl.columns = sl.iloc[0]
sl = sl.iloc[1:]


nx_sl = pd.read_excel("./../../upgrade_mandi/nx/NX Article SL (1).xlsx", "Sheet1")


sl.columns = sl.columns.str.strip()
nx_sl.columns = nx_sl.columns.str.strip()
```

```python
day_of_the_week = [
    "Sunday",
    "Monday",
    "Tuesday",
    "Wednesday",
    "Thursday",
    "Friday",
    "Saturday",
]

# sl.drop(
#     columns=[
#         "Sunday",
#         "Monday",
#         "Tuesday",
#         "Wednesday",
#         "Thursday",
#         "Friday",
#         "Saturday",
#     ]
# )

nx_sl[
    [
        "ITEM_CODE",
    ]
    + day_of_the_week
].columns
# nx_sl.columns
```

```python
formated_sl = pd.concat(
    [
        pd.DataFrame([{col: col for col in sl.columns}]),
        sl.merge(
            nx_sl["ITEM_CODE"],
            how="left_anti",
            left_on="Article Code",
            right_on="ITEM_CODE",
        ).drop(columns=["ITEM_CODE"]),
        (
            nx_sl[
                [
                    "ITEM_CODE",
                ]
                + day_of_the_week
            ]
            .merge(
                sl.drop(columns=day_of_the_week),
                how="inner",
                right_on="Article Code",
                left_on="ITEM_CODE",
            )
            .drop(columns=["ITEM_CODE"])
        ),
    ]
)
```

```python
formated_sl.to_excel(
    "./../../upgrade_mandi/shelf-life/Shelf-Life_30-July-2026.xlsx", index=False
)
```
